## Auxiliary functions

```python
def plot_decision_regions(X, y, classifier, test_idx=None, resolution=0.02):
    
    num2label = {0: "Iris-setosa", 1: "Iris-versicolor", 2: "Iris-virginica"}

    # setup marker generator and color map
    markers = ('s', 'x', 'o', '^', 'v')
    colors = ('red', 'blue', 'lightgreen', 'gray', 'cyan')
    cmap = ListedColormap(colors[:len(np.unique(y))])

    # plot the decision surface
    x1_min, x1_max = X[:, 0].min() - 1, X[:, 0].max() + 1
    x2_min, x2_max = X[:, 1].min() - 1, X[:, 1].max() + 1
    xx1, xx2 = np.meshgrid(np.arange(x1_min, x1_max, resolution),
                           np.arange(x2_min, x2_max, resolution))
    Z = classifier.predict(np.array([xx1.ravel(), xx2.ravel()]).T)
    Z = Z.reshape(xx1.shape)
    plt.contourf(xx1, xx2, Z, alpha=0.4, cmap=cmap)
    plt.xlim(xx1.min(), xx1.max())
    plt.ylim(xx2.min(), xx2.max())

    for idx, cl in enumerate(np.unique(y)):
        plt.scatter(x=X[(y.values == cl).ravel(), 0], y=X[(y.values == cl).ravel(), 1],
                    alpha=0.8, color=cmap(idx),
                    marker=markers[idx], label=num2label.get(cl))
    
    plt.legend()
```

## 0. (Pre)process data
![iris_dataset](../data/img/iris-dataset.png)

```python
X_train = pd.read_csv("../data/iris/X_train.csv", index_col='Id')
y_train = pd.read_csv("../data/iris/y_train.csv", index_col='Id')
X_test = pd.read_csv("../data/iris/X_test.csv", index_col='Id')
y_test = pd.read_csv("../data/iris/y_test.csv", index_col='Id')
```

## 1. Create/Build model - Classification

### Gini index

```python
# from sklearn.tree import DecisionTreeClassifier
classifier_gini = DecisionTreeClassifier(criterion = "gini", # criterium
                                         max_depth = 5, # max tree height
                                         min_samples_leaf = 3, # min leaf samples
                                         random_state = 100)
```

We have different methods
- fit() - fit the modeli.e., learn
- predict() - use learned parameters from .fit() and get predictions
- fit_predict() - does fitting and prediciting over the same dataset (unsupervised learning)

## 2. Fit model
Train the model.

```python
X_train_aux=X_train[['PetalLengthCm', 'PetalWidthCm']]
X_test_aux=X_test[['PetalLengthCm', 'PetalWidthCm']]
```

```python
classifier_gini.fit(X_train_aux, y_train)
```

## 3. Evaluate the model
Make predictions for the test dataset and get some evaluation measurements.

```python
y_pred_gini = classifier_gini.predict(X_test_aux)
```

```python
print("Accuracy: ", accuracy_score(y_test, y_pred_gini)*100)
print("Report: ", classification_report(y_test, y_pred_gini))
```

### Plotting - Visualize the tree

```python
tree.plot_tree(classifier_gini);
```

#### Add some color

```python
fn = ['Sepal length', 'Sepal width', 'Petal length', 'Petal width']
cn = ['Setosa', 'Versicolor', 'Virginica']
fig, axes = plt.subplots(nrows=1, ncols=1)
tree.plot_tree(classifier_gini, feature_names=fn, class_names=cn, filled=True);
```

```python
# If you're satisfied with the model, save it to disk
filename = '../models/classifier_gini.pkl'
pickle.dump(classifier_gini, open(filename, 'wb'))
 
# load the model from disk
# classifier_gini = pickle.load(open(filename, 'rb'))
# result = classifier_gini.predict(X_test)
```

## Linear SVC
## 0. (Pre)process data

```python
# We'll use the petal length and width for this analysis
X_test = X_test[['PetalLengthCm', 'PetalWidthCm']]
X_train = X_train[['PetalLengthCm', 'PetalWidthCm']]
```

### Scale the data
Here we use StandardScaler from sklearn.preprocessing

Standardize features by removing the mean and scaling to unit variance. The standard score of a sample x is calculated as:
\begin{equation}
\Large{z = \frac{x - u}{s}},
\end{equation}
where $u$ and $s$ are the mean and std of the sample, respectively

#### Define the scaler

```python
scaler = StandardScaler()
```

#### Fit the scaler

```python
scaler.fit(X_train)
```

#### Transform the data

```python
X_train_std = scaler.transform(X_train)
X_test_std = scaler.transform(X_test)

print('Dataframe after standardizing features:\n')
print(pd.DataFrame(X_train_std, columns=['PetalLengthCm', 'PetalWidthCm']).head())
```

```python
# save the scaler to disk
filename = '../models/scaler_classification_iris.pkl'
pickle.dump(scaler, open(filename, 'wb'))
 
# load the scaler from disk
# scaler = pickle.load(open(filename, 'rb'))
```

## 1. Create/Build model/classifier

```python
# from sklearn.svm import SVC
svm = SVC(kernel='rbf', random_state=0, gamma=.10, C=10)
```

## 2. Fit model
Train the model.

```python
svm.fit(X_train_std, y_train)
```

## 3. Evaluate the model

```python
print('The accuracy of the svm classifier on training data is {:.2f} out of 1'.format(svm.score(X_train_std, y_train)))
print('The accuracy of the svm classifier on test data is {:.2f} out of 1'.format(svm.score(X_test_std, y_test)))
```

```python
plot_decision_regions(X_test_std, y_test, svm)
```

```python
# If you're satisfied with the model, save it to disk
filename = '../models/classifier_SVC.pkl'
pickle.dump(svm, open(filename, 'wb'))
 
# load the model from disk
# svm = pickle.load(open(filename, 'rb'))
# result = svm.predict(X_test)
```
